```python
import re
import nltk
from nltk.corpus import stopwords
```

```python
print("Upload value:", upload.value)
print("Number of uploaded files:", len(upload.value))
```

```python
stop_words = set(stopwords.words("english"))
def preprocess(text):
    if not text:
        return ""
    
    text = text.lower()
    text = re.sub(r'[^a-zA-Z0-9 ]', ' ', text)
    text = re.sub(r'\s+', ' ', text)
    
    return text.strip()
```

```python
import PyPDF2
import io

def extract_uploaded_resumes(upload_widget):
    resume_texts = []

    for filename in upload_widget.value:
        file_content = upload_widget.value[filename]['content']
        pdf_file = io.BytesIO(file_content)
        reader = PyPDF2.PdfReader(pdf_file)

        text = ""
        for page in reader.pages:
            if page.extract_text():
                text += page.extract_text()

        resume_texts.append(text)

    return resume_texts
```

```python
import pdfplumber
import io

resume_texts = []
resume_names = []

for file in upload.value:
    file_name = file['name']     
    file_content = file['content']

    with pdfplumber.open(io.BytesIO(file_content)) as pdf:
        text = ""
        for page in pdf.pages:
            page_text = page.extract_text()
            if page_text:
                text += page_text

    resume_texts.append(text)
    resume_names.append(file_name)   

print("Extracted resumes:", len(resume_texts))
print("Resume names:", resume_names)
```

```python
cleaned_resumes = [preprocess(resume) for resume in resume_texts]
cleaned_jd = preprocess(jd_box.value)
```

```python
print("Raw resumes count:", len(resume_texts))

cleaned_resumes = [preprocess(r) for r in resume_texts]

print("Cleaned resumes count:", len(cleaned_resumes))
print("First cleaned resume sample:", cleaned_resumes[0][:100])
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

vectorizer = TfidfVectorizer()

all_documents = cleaned_resumes + [cleaned_jd]
tfidf_matrix = vectorizer.fit_transform(all_documents)

resume_vectors = tfidf_matrix[:-1]
jd_vector = tfidf_matrix[-1]

similarity_scores = cosine_similarity(resume_vectors, jd_vector)
```

```python
filtered_resumes = []
filtered_names = []

for text, name in zip(cleaned_resumes, resume_names):
    if text.strip() != "":
        filtered_resumes.append(text)
        
        filtered_names.append(name)

print("Valid resumes:", len(filtered_resumes))
```

```python
cleaned_resumes = filtered_resumes
resume_names = filtered_names
```

```python
from sentence_transformers import SentenceTransformer
model = SentenceTransformer('all-MiniLM-L6-v2')

resume_embeddings = model.encode(cleaned_resumes)
jd_embedding = model.encode([cleaned_jd])

bert_scores = cosine_similarity(resume_embeddings, jd_embedding)
```

```python
bert_scores 
```

```python
required_skills = [
    "Python", "Machine Learning", "Deep Learning",
    "sql", "Pandas", "Numpy",
    "Scikit-Learn", "NLP", "Data Analysis", "TensorFlow"
]
```

```python
def skill_match(resume_text, skills):
    matched = []
    for skill in skills:
        if skill in resume_text:
            matched.append(skill)
    
    percentage = (len(matched) / len(skills)) * 100
    missing = list(set(skills) - set(matched))
    
    return percentage, matched, missing
```

```python
def extract_experience(text):
    matches = re.findall(r'(\d+)\s+years', text.lower())
    if matches:
        return max([int(m) for m in matches])
    return 0
```

```python
import pandas as pd
```

```python
results = []

for i in range(len(cleaned_resumes)):
    
    tfidf_score =similarity_scores[i][0] * 100
    bert_score = bert_scores[i][0] * 100
    
    skill_score, matched, missing = skill_match(
        cleaned_resumes[i], required_skills
    )
    
    experience_years = extract_experience(resume_texts[i])
    experience_score = min(experience_years*5, 10 )  # Max 10%
    
    final_score = (
        0.5 * bert_score +
        0.2 * tfidf_score +
        0.3 * skill_score +
         experience_score
    )
    
    results.append({
        "Resume Name": resume_names[i],
        "BERT Similarity (%)": round(bert_score,2),
        "TFIDF Similarity (%)": round(tfidf_score,2),
        "Skill Match (%)": round(skill_score,2),
        "Experience (Years)": experience_years,
        "Matched Skills": ", ".join(matched),
        "Missing Skills": ", ".join(missing),
        "Final ATS Score (%)": round(final_score,2)
    })

results_df = pd.DataFrame(results)
results_df = results_df.sort_values(
    by="Final ATS Score (%)", ascending=False
)

results_df
```

```python
top_resumes = results_df[results_df["Final ATS Score (%)"] >= 60]
top_resumes
```

```python
def label(score):
    if score >= 80:
        return "Excellent Fit"
    elif score >= 60:
        return "Good Fit" 
    elif score >= 50:
        return "Average Fit"
    else:
        return "Low Match"

results_df["Category"] = results_df["Final ATS Score (%)"].apply(label)
```
